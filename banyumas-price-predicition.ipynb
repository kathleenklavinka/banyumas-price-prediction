```python
komoditas = ['Beras', 'Daging Ayam', 'Daging Sapi', 'Telur Ayam',
             'Bawang Merah', 'Bawang Putih', 'Cabai Merah', 'Cabai Rawit',
             'Minyak Goreng', 'Gula Pasir']
```

```python
print(pgn.columns.tolist())
```

```python
pgn['Tahun'] = pd.to_numeric(pgn['Tahun'], errors='coerce')
```

```python
# Menghitung Korelasi Pearson berdasrakan rumus di paper
def hitung_korelasi_pearson(x, y):
    x = np.array(x, dtype=float)
    y = np.array(y, dtype=float)
    n = len(x)

    sum_x = np.sum(x)
    sum_y = np.sum(y)
    sum_xy = np.sum(x * y)
    sum_x2 = np.sum(x ** 2)
    sum_y2 = np.sum(y ** 2)

    denom_x = n * sum_x2 - sum_x**2
    denom_y = n * sum_y2 - sum_y**2

    # Cek variansi nol
    if denom_x <= 0 or denom_y <= 0:
        return 0.0

    pembilang = n * sum_xy - sum_x * sum_y
    penyebut = np.sqrt(denom_x * denom_y)

    return pembilang / penyebut


def matriks_korelasi_pearson(dataframe):
    dataframe = dataframe.select_dtypes(include=[np.number]) # Agar hanya menghtiung angka
    kolom = dataframe.columns
    n_kolom = len(kolom)
    matriks = np.zeros((n_kolom, n_kolom))

    for i in range(n_kolom):
        for j in range(n_kolom):
            if i == j:
                matriks[i, j] = 1.0
            else:
                matriks[i, j] = hitung_korelasi_pearson(
                    dataframe.iloc[:, i],
                    dataframe.iloc[:, j]
                )

    return pd.DataFrame(matriks, index=kolom, columns=kolom)

# Hitung Korelasi Pearson
pgn = pgn.drop(columns=['Tahun']) # Agar tahun tidak ikut dihitung ke dalam korelasi
korelasi_pearson = matriks_korelasi_pearson(pgn)

# Print hasil Korelasi Pearson
print(korelasi_pearson.round(2))

# Plot heatmap untuk Korelasi Pearson
plt.figure(figsize=(10, 8))
sns.heatmap(korelasi_pearson, annot=True, cmap='RdBu',
            center=0, square=True, fmt='.2f', cbar_kws={'shrink': .8})
plt.title('Heatmap Korelasi Pearson antar Komoditas')
plt.tight_layout()
plt.show()
```

```python
# Perhitungan korelasi Pearson menggunakan library (sama seperti diatas)
# Ambil subset dataframe hanya untuk kolom-kolom komoditas
pgn2 = pgn[komoditas]

# Hitung matriks korelasi Pearson
korelasi = pgn2.corr(method='pearson')

# Tampilkan
print(korelasi)
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(korelasi, annot=True, cmap='RdYlBu_r', fmt=".2f", linewidths=0.5)
plt.title('Heatmap Korelasi Pearson antar Komoditas')
plt.show()
```
